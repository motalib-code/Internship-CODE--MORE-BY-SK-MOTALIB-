# Exploratory Data Analysis — Titanic Dataset

**Objective:** Perform EDA on the Titanic passenger dataset to understand patterns, correlations, and trends that influenced survival.

## 1. Setup & Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["figure.figsize"] = (10, 6)
sns.set_style("whitegrid")
print("Libraries loaded successfully")
```

## 2. Load Cleaned Data

```python
df = pd.read_csv("../data/cleaned_dataset.csv")
print(f"Shape: {df.shape}")
df.head()
```

```python
df.info()
```

## 3. Summary Statistics

```python
df.describe()
```

```python
# Survival rate
survival_rate = df["Survived"].mean() * 100
print(f"Overall survival rate: {survival_rate:.2f}%")

# By class
print("\nSurvival rate by passenger class:")
print(df.groupby("Pclass")["Survived"].mean() * 100)

# By sex
print("\nSurvival rate by sex:")
print(df.groupby("Sex")["Survived"].mean() * 100)
```

## 4. Missing Value Check

```python
print("Missing values after cleaning:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

## 5. Visualization 1 — Age Distribution

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(df["Age"], bins=30, kde=True, ax=ax[0], color="steelblue")
ax[0].set_title("Age Distribution", fontsize=14, fontweight="bold")
ax[0].set_xlabel("Age")
ax[0].set_ylabel("Count")

sns.boxplot(x="Survived", y="Age", data=df, ax=ax[1], palette="Set2")
ax[1].set_title("Age vs Survival", fontsize=14, fontweight="bold")
ax[1].set_xlabel("Survived (0 = No, 1 = Yes)")
ax[1].set_ylabel("Age")

plt.tight_layout()
plt.savefig("../images/age_histogram.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: images/age_histogram.png")
```

## 6. Visualization 2 — Fare by Passenger Class

```python
fig, ax = plt.subplots(figsize=(10, 6))
sns.boxplot(x="Pclass", y="Fare", hue="Survived", data=df, ax=ax, palette="muted")
ax.set_title("Fare Distribution by Passenger Class and Survival", fontsize=14, fontweight="bold")
ax.set_xlabel("Passenger Class")
ax.set_ylabel("Fare ($)")
ax.legend(title="Survived", labels=["No", "Yes"])
plt.ylim(0, 150)
plt.tight_layout()
plt.savefig("../images/fare_vs_class.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: images/fare_vs_class.png")
```

## 7. Visualization 3 — Correlation Heatmap

```python
numeric_cols = ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]
corr_matrix = df[numeric_cols].corr()

fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f",
            linewidths=0.5, square=True, ax=ax,
            cbar_kws={"shrink": 0.8})
ax.set_title("Feature Correlation Matrix", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("../images/correlation_heatmap.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: images/correlation_heatmap.png")
```

## 8. Additional Analysis — Survival by Gender & Class

```python
fig, ax = plt.subplots(figsize=(10, 6))
ct = pd.crosstab([df["Sex"], df["Pclass"]], df["Survived"])
ct_pct = ct.div(ct.sum(axis=1), axis=0) * 100
ct_pct.plot(kind="bar", stacked=True, ax=ax, color=["#ff6b6b", "#51cf66"])
ax.set_title("Survival Rate by Gender and Passenger Class", fontsize=14, fontweight="bold")
ax.set_xlabel("(Gender, Class)")
ax.set_ylabel("Percentage")
ax.legend(title="Survived", labels=["No", "Yes"])
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../images/survival_gender_class.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: images/survival_gender_class.png")
```

## 9. Key Insights

- **Survival rate:** ~38% of passengers survived the disaster.
- **Gender effect:** Women had a significantly higher survival rate (~74%) than men (~19%).
- **Class effect:** 1st class passengers survived at ~63%, while 3rd class survived at ~24%.
- **Age:** Children (0–10) had higher survival; passengers aged 20–40 formed the largest group.
- **Fare:** Higher fares (1st class) correlated with better survival outcomes.
- **Family:** Passengers with 1–2 family members (SibSp/Parch) had slightly better survival.

### Correlation Highlights

- `Survived` — negatively correlated with `Pclass` (-0.34): lower class → lower survival.
- `Survived` — positively correlated with `Fare` (0.26): higher fare → higher survival.
- `Pclass` and `Fare` are strongly correlated (-0.55): 1st class paid more.
